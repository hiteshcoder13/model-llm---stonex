```python
import requests
import time

# 🔐 आपका Bearer Token यहाँ डालो
API_KEY = "30a5cf47da51fc851867d87c4ea20975a15f4568"

BASE_URL = "https://app.stonexgalleria.com/ai"
LIMIT = 1000  # per page
```

```python
def fetch_page(offset):
    url = f"{BASE_URL}?limit={LIMIT}&offset={offset}"
    
    headers = {
        "Authorization": f"Bearer {API_KEY}"
    }
    
    response = requests.get(url, headers=headers)
    
    if response.status_code != 200:
        print(f"❌ Error: {response.status_code}")
        return None
    
    return response.json()
```

```python
TARGET_FAMILY = "VIBRANIUM"

offset = 0
found = False
results = []

while True:
    print(f"🔄 Fetching offset: {offset}")
    
    data = fetch_page(offset)
    
    if not data or "data" not in data or "lots" not in data["data"]:
        print("⚠️ Invalid response or no more data")
        break
    
    lots = data["data"]["lots"]
    
    # अगर data empty है → end of pagination
    if not lots:
        print("🚫 No more data available")
        break
    
    for lot in lots:
        family = lot.get("family", "").strip().upper()
        
        if family == TARGET_FAMILY:
            print("✅ FOUND VIBRANIUM!")
            results.append(lot)
            found = True
    
    # अगर मिल गया तो loop break कर सकते हो
    if found:
        break
    
    offset += LIMIT
    
    # ⚠️ optional: rate limit safe
    time.sleep(0.5)
```

```python
if results:
    print(f"\n🎯 Found {len(results)} VIBRANIUM entries:\n")
    
    for i, lot in enumerate(results, 1):
        print(f"--- Result {i} ---")
        print("Lot No:", lot.get("lotno"))
        print("Family:", lot.get("family"))
        print("BG Image:", lot.get("bg_removed_image"))
        
        slabs = lot.get("slabs", [])
        if slabs:
            print("Slab Image:", slabs[0].get("framed_stand_img_1"))
        
        print()
else:
    print("❌ VIBRANIUM not found in dataset")
```
